```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams.update(
    {
        "font.size": 11,  # base font size (tick labels)
        "axes.labelsize": 13,  # axis labels
        "axes.titlesize": 13,
        "legend.fontsize": 11,
        "xtick.labelsize": 11,
        "ytick.labelsize": 11,
        "figure.dpi": 300,
    }
)
```

```python
# --- Files to load ---
files = {
    "3D": "lo_distr_overlaps_hopdc_3D.npz",
    "4D": "lo_distr_overlaps_hopdc_4D.npz",
    "5D": "lo_distr_overlaps_hopdc_5D.npz",
    # Add more if needed
}
```

```python
fig, ax = plt.subplots(figsize=(3.4, 2.6))

for label, filename in files.items():
    data = np.load(filename)
    taup = data["taup"]
    tauf = data["tauf"]
    overlap = data["overlap"]
    infidelity = data["infidelity"]

    ax.plot((taup**2 / tauf**2), overlap**2, markersize=4, label=label, marker="o")


ax.set_xlabel(r"$\beta_{2P}/\beta_{2F}$")
ax.set_ylabel(r"$\mathcal{F}$")
ax.legend()
ax.grid(True, which="both", alpha=0.3)
ax.text(-0.15, 1.05, "(b)", transform=ax.transAxes, fontsize=12, va="bottom", ha="left")

plt.xlim([0, 23])
plt.tight_layout()
plt.savefig(f"lo_distr_hopdc_fidelity_general.pdf", dpi=300, bbox_inches="tight")

plt.show()
```
